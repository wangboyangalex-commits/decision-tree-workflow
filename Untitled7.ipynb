```python
# =========================
# 1. Load Data & Define Target
# =========================

import pandas as pd
import numpy as np

# 读取事件级数据：一行 = 一次加购事件
df = pd.read_csv("事件级表.csv")

# 把时间转换成 datetime，后面用于时间切分
df["timestamp"] = pd.to_datetime(df["timestamp"])

# 创建目标变量：
# 1 = 7天内没有购买
# 0 = 7天内购买了
df["non_conversion_7d"] = 1 - df["converted_7d"]

# 简单检查数据没有读错
assert len(df) == 20000
assert set(df["converted_7d"].unique()) == {0, 1}

print("Data shape:", df.shape)
print("7-day non-conversion rate:", round(df["non_conversion_7d"].mean(), 4))
```

```python
# =========================
# 2. Prepare Train / Test Data
# =========================

# 【项目适配】按时间切分：
# 较早事件训练，10/24–10/31 留作 7 天 Gap，11 月作为最终测试
train = df[df["timestamp"] < "2022-10-24"].copy()
test = df[(df["timestamp"] >= "2022-11-01") & (df["timestamp"] < "2022-12-01")].copy()

# 不进入模型的字段：
# ID / 时间 / 商品编号 / 结果变量 / add_month
drop_cols = [
    "client_id", "timestamp", "sku", "add_month",
    "converted_1d", "converted_3d",
    "converted_7d", "converted_14d",
    "non_conversion_7d"]


# 从剩余变量中，把 category 和 customer_segment 单独拿出来做分类编码
numeric_features = [
    col for col in df.columns
    if col not in drop_cols + ["category", "customer_segment"]
]


# 【课件对应：Missing Values】
# NULL 表示没有对应历史行为，因此统一用 -1 保留这个信息
X_train = train[numeric_features].copy().fillna(-1)
X_test = test[numeric_features].copy().fillna(-1)


# 【课件对应：Categorical Encoding】
# category 类别太多，只保留 Train 中最常见的 15 类
# 其他类别统一归为 OTHER，避免维度爆炸
top_categories = train["category"].value_counts().head(15).index

X_train["category_group"] = train["category"].where(
    train["category"].isin(top_categories), "OTHER"
).astype(str)

X_test["category_group"] = test["category"].where(
    test["category"].isin(top_categories), "OTHER"
).astype(str)


# 加入 customer_segment
X_train["customer_segment"] = train["customer_segment"].astype(str)
X_test["customer_segment"] = test["customer_segment"].astype(str)


# 把两个分类变量转成 Dummy Variables
X_train = pd.get_dummies(
    X_train,
    columns=["category_group", "customer_segment"],
    dtype=int
)

X_test = pd.get_dummies(
    X_test,
    columns=["category_group", "customer_segment"],
    dtype=int
)


# 保证 Test 和 Train 拥有完全相同的 Predictor
X_test = X_test.reindex(columns=X_train.columns, fill_value=0)


# 建立目标变量
y_train = train["non_conversion_7d"]
y_test = test["non_conversion_7d"]


# 简单检查最终数据
print("Train:", X_train.shape)
print("Test:", X_test.shape)
print("Columns matched:", list(X_train.columns) == list(X_test.columns))
```

```python
# =========================
# 3. Full Decision Tree & Overfitting Check
# =========================

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

# 【课件对应：Decision Tree P47–48】
# 先建立一棵不限制深度的完整树，检查是否过拟合
full_tree = DecisionTreeClassifier(
    criterion="gini",   # 使用 Gini 选择最佳分裂
    random_state=42     # 固定结果，保证可重复
)

# 用 Train 训练模型
full_tree.fit(X_train, y_train)

# 分别预测 Train 和 Test
train_pred = full_tree.predict(X_train)
test_pred = full_tree.predict(X_test)

# 比较 Train / Test 表现，并查看树复杂度
print("Train Accuracy:", round(accuracy_score(y_train, train_pred), 4))
print("Test Accuracy:", round(accuracy_score(y_test, test_pred), 4))
print("Tree Depth:", full_tree.get_depth())
print("Leaves:", full_tree.get_n_leaves())
```

```python
# =========================
# 4. Forward Validation & Hyperparameter Tuning
# =========================

from sklearn.metrics import roc_auc_score

# 【项目适配】每一轮都用过去训练、未来验证，中间留 7 天 Gap
folds = [
    ("2022-07-25", "2022-08-01", "2022-09-01"),
    ("2022-08-25", "2022-09-01", "2022-10-01"),
    ("2022-09-24", "2022-10-01", "2022-10-24")
]

# 只测试老师课件 P49 讲过的两个控制过拟合参数
depth_list = [2, 3, 4]
leaf_list = [100, 200]

results = []

# 比较不同 max_depth 和 min_samples_leaf
for depth in depth_list:
    for leaf in leaf_list:

        auc_list = []

        for train_before, valid_from, valid_before in folds:

            # 按时间建立当前 Fold 的 Train / Validation
            tr = df[df["timestamp"] < train_before].copy()
            va = df[
                (df["timestamp"] >= valid_from) &
                (df["timestamp"] < valid_before)
            ].copy()

            # 缺失值：无历史记为 -1
            X_tr = tr[numeric_features].copy().fillna(-1)
            X_va = va[numeric_features].copy().fillna(-1)

            # category 只根据当前 Train 选 Top 15，防止偷看未来
            top_cat = tr["category"].value_counts().head(15).index

            X_tr["category_group"] = tr["category"].where(
                tr["category"].isin(top_cat), "OTHER"
            ).astype(str)

            X_va["category_group"] = va["category"].where(
                va["category"].isin(top_cat), "OTHER"
            ).astype(str)

            X_tr["customer_segment"] = tr["customer_segment"].astype(str)
            X_va["customer_segment"] = va["customer_segment"].astype(str)

            # 分类变量编码
            X_tr = pd.get_dummies(
                X_tr,
                columns=["category_group", "customer_segment"],
                dtype=int
            )

            X_va = pd.get_dummies(
                X_va,
                columns=["category_group", "customer_segment"],
                dtype=int
            )

            X_va = X_va.reindex(columns=X_tr.columns, fill_value=0)

            # 建立当前候选树
            model = DecisionTreeClassifier(
                criterion="gini",
                max_depth=depth,
                min_samples_leaf=leaf,
                random_state=42
            )

            # 训练并计算 Validation ROC-AUC
            model.fit(X_tr, tr["non_conversion_7d"])
            score = model.predict_proba(X_va)[:, 1]

            auc_list.append(
                roc_auc_score(va["non_conversion_7d"], score)
            )

        # 保存三轮 Validation 的平均 AUC
        results.append([
            depth,
            leaf,
            np.mean(auc_list)
        ])


# 整理结果，AUC 越高越好
cv_results = pd.DataFrame(
    results,
    columns=["max_depth", "min_samples_leaf", "Mean_AUC"]
).sort_values("Mean_AUC", ascending=False)

print(cv_results.round(4))
```

```python
# =========================
# 5. Final Decision Tree & Evaluation
# =========================

from sklearn.metrics import (
    accuracy_score, precision_score,
    recall_score, f1_score,
    roc_auc_score, confusion_matrix
)

# 从 Validation 结果中自动取得最佳参数
best_depth = int(cv_results.iloc[0]["max_depth"])
best_leaf = int(cv_results.iloc[0]["min_samples_leaf"])

# 【课件对应：P49 Pruning】
# 用最佳 max_depth 和 min_samples_leaf 控制树复杂度
final_tree = DecisionTreeClassifier(
    criterion="gini",
    max_depth=best_depth,
    min_samples_leaf=best_leaf,
    random_state=42
)

# 用完整 Train 数据训练最终模型
final_tree.fit(X_train, y_train)

# 对 11 月 Test 做预测
y_pred = final_tree.predict(X_test)

# 得到每条事件“7天内不购买”的预测概率
risk_score = final_tree.predict_proba(X_test)[:, 1]

# 【课件对应：P79 Evaluation Metrics】
# 计算最终模型表现
print("Best max_depth:", best_depth)
print("Best min_samples_leaf:", best_leaf)

print("\nAccuracy:", round(accuracy_score(y_test, y_pred), 4))
print("Precision:", round(precision_score(y_test, y_pred), 4))
print("Recall:", round(recall_score(y_test, y_pred), 4))
print("F1:", round(f1_score(y_test, y_pred), 4))
print("ROC-AUC:", round(roc_auc_score(y_test, risk_score), 4))

# 查看 Confusion Matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# 【项目要求：与朴素基线比较】
# 假设所有事件都预测为“7天内不购买”
naive_pred = np.ones(len(y_test), dtype=int)

print(
    "\nNaive Baseline Accuracy:",
    round(accuracy_score(y_test, naive_pred), 4)
)
```

```python
# =========================
# 6. Top 10% Risk Lift
# =========================

# 把 Test 的真实结果和模型风险分数放在一起
lift_df = pd.DataFrame({
    "actual": y_test.values,
    "risk_score": risk_score
})

# 按“7天内不购买”的风险从高到低排序
lift_df = lift_df.sort_values(
    "risk_score",
    ascending=False
)

# 取风险最高的前 10%
top_n = int(len(lift_df) * 0.10)
top_10 = lift_df.head(top_n)

# 整体未转化率
overall_rate = lift_df["actual"].mean()

# Top 10% 高风险事件的实际未转化率
top10_rate = top_10["actual"].mean()

# Lift = Top 10% 未转化率 ÷ 整体未转化率
lift = top10_rate / overall_rate

print("Overall non-conversion rate:", round(overall_rate, 4))
print("Top 10% non-conversion rate:", round(top10_rate, 4))
print("Top 10% Lift:", round(lift, 4))
```

```python
# =========================
# 7. Check Ties at Top 10% Cutoff
# =========================

# 找到 Top 10% 最后一条记录对应的风险分数
cutoff_score = top_10["risk_score"].min()

# 统计风险分数严格高于 cutoff 的事件数
higher_count = (lift_df["risk_score"] > cutoff_score).sum()

# 统计刚好等于 cutoff 的事件数
tie_count = (lift_df["risk_score"] == cutoff_score).sum()

# 输出结果
print("Top 10% cutoff score:", round(cutoff_score, 4))
print("Events above cutoff:", higher_count)
print("Events tied at cutoff:", tie_count)
print("Target Top 10% size:", top_n)
```

```python
# =========================
# 8. Tie-adjusted Top 10% Lift
# =========================

# Top 10% 还需要从并列组中取多少条
needed_from_ties = top_n - higher_count

# 风险严格高于 cutoff 的事件，实际未转化总数
higher_actual = lift_df.loc[
    lift_df["risk_score"] > cutoff_score,
    "actual"
].sum()

# cutoff 并列组的平均实际未转化率
tie_rate = lift_df.loc[
    lift_df["risk_score"] == cutoff_score,
    "actual"
].mean()

# 用并列组的平均表现估计剩余名额
adjusted_top10_actual = (
    higher_actual +
    needed_from_ties * tie_rate
)

# 调整后的 Top 10% 未转化率
adjusted_top10_rate = adjusted_top10_actual / top_n

# 调整后的 Lift
adjusted_lift = adjusted_top10_rate / overall_rate

print("Adjusted Top 10% non-conversion rate:",
      round(adjusted_top10_rate, 4))

print("Adjusted Top 10% Lift:",
      round(adjusted_lift, 4))
```

```python
# =========================
# 9. Feature Importance
# =========================

# 提取最终决策树中每个 Predictor 的重要性
importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": final_tree.feature_importances_
})

# 按重要性从高到低排序，只看最重要的前 10 个
importance = importance.sort_values(
    "Importance",
    ascending=False
).head(10)

# 显示结果
print(importance)
```
